```python
import pandas as pd
import numpy as np
```

```python
#reading data set in csv format as df
df=pd.read_csv("/content/student_bigdata.csv")
```

```python
#displaying top 5 rows
df.head(5)
```

```python
# data shape - displays the number of columns and rows
df.shape
```

```python
#displays the column names
df.columns
```

```python
#finding the number of null entries in each column
df.isnull().sum()
```

```python
#checking the data types
df.dtypes
```

```python
#sorting the numerical columns from categorical columns and printing mean median and standard deviation
num_cols = df.select_dtypes(include=['number']).columns

for col in num_cols:
  print(f"Column : {col}")
  mean=df[f'{col}'].mean()
  median=df[f'{col}'].median()
  std=df[f'{col}'].std()
  print(f"Mean :{mean}")
  print(f"Median :{median}")
  print(f"Standard Deviation :{std}")
  print(" ____ ")
  print("")
```

```python
#plotting three graphs
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

#histogram number of student vs math score with 10 bins
scores = df['Math_Score']

plt.figure()
bins = np.linspace(df['Math_Score'].min(), df['Math_Score'].max(), 10)

plt.hist(scores, bins=bins,linewidth=2,histtype='stepfilled', label='student count')


plt.xlabel('Maths score')
plt.ylabel('Number of Students')
plt.title('Math Score Distribution')
plt.legend()
plt.show()
#this histogram shows that very less number of students scored less than 40 and the minimum score is 30 .highest concentration of marks lie around 75 to 85.this histogram shows fairly normal distribution of maths marks.
```

```python
plt.figure()

plt.scatter(df['Study_Hours'], df['Science_Score'])

plt.xlabel('Study Hours')
plt.ylabel('Science_Scores')
plt.title('Study Hours vs Science Scores')
plt.show()

# there is no definite trend or relation of study hours vs science scores. this shows that even though if a student had studied more number of hours this is not reflected in their science marks
```

```python
# Count Final_Result for each Gender
gender_result = df.groupby('Gender')['Final_Result'].value_counts().unstack()

plt.figure()

for gender in gender_result.index:
    plt.pie(gender_result.loc[gender],
            labels=gender_result.columns,
            autopct='%1.1f%%',
            startangle=90)
    plt.title(f'Final Result Distribution - {gender}')
    plt.show()
    #pass percentage in male is slightly more than that of female and both are around 75%
```
